```python
name = "mypackage"
num = 99
size = 10.123456789098765432

base = "A package named {name} with {num} items of size={size:.2f}"
f_string = f"A package named {name} with {num} items of {size=:.2f}"
t_string = t"A package named {name} with {num} items of {size=:.2f}"

for s in base, f_string, t_string:
    print(type(s))
    print(s)
```

```python
from string import templatelib

assert isinstance(t_string, templatelib.Template)
dir(templatelib.Template)
```

```python
print(type(t_string.strings))
print(type(t_string.strings[0]))
assert all(isinstance(s, str) for s in t_string.strings)
print(len(t_string.strings))
print(t_string.strings)
```

```python
print(t_string.values)
```

```python
print(type(t_string.interpolations))
print(type(t_string.interpolations[0]))
assert all(isinstance(s, templatelib.Interpolation) for s in t_string.interpolations)
print(len(t_string.interpolations))
for i in t_string.interpolations:
    print(i)
```

```python
dir(templatelib.Interpolation)
```

```python
display([i.conversion for i in t_string.interpolations])
display([i.expression for i in t_string.interpolations])
display([i.format_spec for i in t_string.interpolations])
display([i.value for i in t_string.interpolations])
```

```python
display([type(i) for i in t_string])
display(list(t_string))
```

```python
from tstrings import t_string_to_str # local file

print(base.format(name=name, num=num, size=size))
print(f_string)
print(t_string_to_str(t_string))
```
